```python
from collections import deque

def flood_fill_maze(rows, cols, walls, goal):
    """
    Simulates the flood fill algorithm on a grid maze.
    Calculates the shortest distance from every cell to the goal.
    """
    # Initialize distance grid with infinity (-1 represents unvisited)
    distances = [[-1 for _ in range(cols)] for _ in range(rows)]
    
    # Queue for Breadth-First Search (BFS) flood fill
    queue = deque([goal])
    distances[goal[0]][goal[1]] = 0
    
    # Possible movements: Up, Down, Left, Right
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    
    while queue:
        r, c = queue.popleft()
        
        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            
            # Check grid boundaries
            if 0 <= nr < rows and 0 <= nc < cols:
                # Check if there is a wall between current cell and neighbor
                # walls is a set of tuples representing blocked edges: ((r1, c1), (r2, c2))
                if ((r, c), (nr, nc)) not in walls and ((nr, nc), (r, c)) not in walls:
                    if distances[nr][nc] == -1:  # If unvisited
                        distances[nr][nc] = distances[r][c] + 1
                        queue.append((nr, nc))
                        
    return distances

def print_grid(grid):
    for row in grid:
        print(" ".join(f"{val:2}" if val != -1 else " . " for val in row))

# Define a 4x4 maze configuration
ROWS, COLS = 4, 4
GOAL = (0, 3) # Top-right corner

# Define walls as bidirectional blocked edges
# Format: ((r1, c1), (r2, c2))
maze_walls = {
    ((0, 0), (0, 1)),
    ((1, 0), (2, 0)),
    ((2, 1), (2, 2)),
    ((1, 2), (1, 3)),
}

# Run flood fill
distance_matrix = flood_fill_maze(ROWS, COLS, maze_walls, GOAL)

print("Flood Fill Distance Matrix (Goal at top-right):")
print_grid(distance_matrix)
```

```python
# 4x4 maze itself 
def print_ascii_maze(rows, cols, walls, goal):
    """
    Renders the 4x4 maze visually using ASCII characters,
    showing open paths, walls, and the goal position.
    """
    # Top boundary of the maze
    print("+" + "---+" * cols)
    
    for r in range(rows):
        row_str = "|"
        for c in range(cols):
            # Mark the goal cell with 'G', otherwise empty space
            cell_content = " G " if (r, c) == goal else "   "
            
            # Check if there is a right wall between (r, c) and (r, c+1)
            has_right_wall = (
                c == cols - 1 or 
                ((r, c), (r, c + 1)) in walls or 
                ((r, c + 1), (r, c)) in walls
            )
            row_str += cell_content + ("|" if has_right_wall else " ")
        print(row_str)
        
        # Bottom boundary for the current row
        bottom_str = "+"
        for c in range(cols):
            has_bottom_wall = (
                r == rows - 1 or 
                ((r, c), (r + 1, c)) in walls or 
                ((r + 1, c), (r, c)) in walls
            )
            bottom_str += ("---+" if has_bottom_wall else "   +")
        print(bottom_str)

# Visualizing the 4x4 maze structure used previously
ROWS, COLS = 4, 4
GOAL = (0, 3)
maze_walls = {
    ((0, 0), (0, 1)),
    ((1, 0), (2, 0)),
    ((2, 1), (2, 2)),
    ((1, 2), (1, 3)),
}

print("Visual Layout of the 4x4 Maze:")
print_ascii_maze(ROWS, COLS, maze_walls, GOAL)
```
